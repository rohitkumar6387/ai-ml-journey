```python
import os
from groq import Groq

client = Groq(api_key = "enter api key")
chat_completion = client.chat.completions.create(
    messages = [
        {
            "role":"user",
            "content":"Explain quantum computing",
        }
    ],
    model = "qwen/qwen3.6-27b",
)

print(chat_completion.choices[0].message.content)
```

```python
chat_completion = client.chat.completions.create(
    messages = [
        {
            "role":"user","content":"a+2=7,find a"
        },
        {
            "role":"system","content":"act like a math teacher"
        }
    ],
    model = "qwen/qwen3.6-27b",
)

print(chat_completion.choices[0].message.content)
```

```python
#Images
client = Groq(api_key="enter api key")
chat_completion = client.chat.completions.create(
    model = "qwen/qwen3.6-27b",
    messages = [
        {
            "role":"user",
            "content":[
                {"type":"text",
                 "text":"Tell me capital of the country and the prime minister name to which the flag belong to in the reference image."
                },
                {"type":"image_url",
                 "image_url":{"url":"https://static.india.com/wp-content/uploads/2023/07/2246019dd9c3e828383daa343e8a7a25.jpg?impolicy=Medium_Widthonly&w=400&h=800"
                }
                }
                    ]
                    
        }
    ],
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=False,
    stop=None,
    
)

print(chat_completion.choices[0].message)
```
